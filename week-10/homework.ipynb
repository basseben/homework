**Task:**

Create a scatter plot that produces a spiral/cyclone pattern.

* Generate radius values that gradually increase
* Generate angle values that rotate around the origin
* Compute:
  * x = radius × cos(angle)
  * y = radius × sin(angle)
* Plot the points using plt.scatter()
* Add a title

````python
# Your Solution
# Radius values that gradually increase
radius = np.linspace(0, 10, 1000)

# Angle values that rotate around the origin
theta = np.linspace(0, 15 * np.pi, 1000)

#  Compute the x-coordinate by cosine of its angle
x = radius * np.cos(theta)

# Compute y-coordinate bys sine of its angle
y = radius * np.sin(theta)

# 5. Scatter plot plus title
plt.figure(figsize=(5,5))
plt.scatter(x, y)
plt.title("Spiral Pattern")
plt.show()
None
````

**Review Problem 2**

**Concept:** *reading CSV data, matplotlib, line graph*

**Task:**

Write a function `read_and_plot_column(filename, column_name)` that:
* Opens a CSV file (do **not** use pandas)
* Extracts the values from the specified column
* Casts all values to `float`
* Plots the values using `plt.plot()`
* Adds a title
* Displays the plot

````python
# Your Solution
def read_and_plot_column(filename, column_name):
    """Open a CSV file, extract a column, cast to float, and plot it."""
    with open(filename) as fp:
        lines = fp.read().strip().split("\n")

    headers = [h.strip() for h in lines[0].split(",")]

    if column_name not in headers:
        raise ValueError(
            f"Column '{column_name}' not found. Available columns: {headers}"
        )

    col_index = headers.index(column_name)

    # Extracts the values from the specified column, casting to float
    values = []
    for line in lines[1:]:
        fields = line.split(",")
        values.append(float(fields[col_index]))

    # Plots the values using plt.plot()
    plt.plot(values)

    plt.title(f"{column_name} from {filename}")

    plt.show()
````
